```python
import pandas as pd
from alphaQuantSystem.data import DataEngine
de = DataEngine()
df = de.get_hist_data("159509.SZ", period="D", start_date="20230101", end_date="20260701")
df = df.copy()
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df = df.dropna(subset=["date"]).sort_values("date").reset_index(drop=True)
all_trade_date = df["date"]
```

```python
df_list = []
for date in all_trade_date:
    # 打印年月日字符串
    print(date.strftime("%Y-%m-%d"))
    # 方式A：传date对象（大多数接口支持）
    df = wufu_momentum_score.score_etf_pool([symbol], date.strftime("%Y-%m-%d"))
    # 方式B：传字符串 "2023-08-08"
    # df = wufu_momentum_score.score_etf_pool([symbol], date.strftime("%Y-%m-%d"))
    df_list.append(df)

mometum_result = pd.concat(df_list, ignore_index=True)
```
